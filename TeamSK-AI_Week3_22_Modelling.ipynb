**Dataset:** Pima Indians Diabetes Database  
**Problem Type:** Binary Classification  
**Target Variable:** Outcome (0 = No Diabetes, 1 = Diabetes)

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('diabetes.csv')
print(df.shape)
print(df.head())
print(df.isnull().sum())
```

```python
X = df.drop('Outcome', axis=1)
y = df['Outcome']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print(f'Train size: {X_train.shape[0]}')
print(f'Test size:  {X_test.shape[0]}')
```

## Model 1: Logistic Regression (Baseline)
Logistic Regression is our baseline model. It is the simplest linear classifier and serves as the benchmark for all other models.

```python
lr = LogisticRegression(random_state=42)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
```

```python
print('=== Logistic Regression — Test Set Results ===')
print(classification_report(y_test, y_pred_lr))
lr_accuracy = accuracy_score(y_test, y_pred_lr)
```

```python
cv_scores_lr = cross_val_score(lr, X_scaled, y, cv=5, scoring='accuracy')
print(f'LR CV Scores: {cv_scores_lr}')
print(f'LR CV Mean:   {cv_scores_lr.mean():.4f}')
print(f'LR CV Std:    {cv_scores_lr.std():.4f}')
```

## Model 2: Decision Tree
Decision Tree is a non-linear model that splits data by feature 
thresholds. Unlike Logistic Regression, it can capture non-linear 
boundaries in the data.



```python
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
```

```python
print("=== Decision Tree — Test Set Results ===")
print(classification_report(y_test, y_pred_dt))
dt_accuracy = accuracy_score(y_test, y_pred_dt)
```

```python
cv_scores_dt = cross_val_score(dt, X_scaled, y, cv=5, scoring='accuracy')
print(f"DT Cross-Validation Scores: {cv_scores_dt}")
print(f"DT CV Mean: {cv_scores_dt.mean():.4f}")
print(f"DT CV Std:  {cv_scores_dt.std():.4f}")
```

## Model 3: K-Nearest Neighbors (KNN)
KNN classifies a sample based on the majority class of its k nearest 
neighbors. It is a distance-based, non-parametric model.

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
```

```python
print("=== KNN — Test Set Results ===")
print(classification_report(y_test, y_pred_knn))
knn_accuracy = accuracy_score(y_test, y_pred_knn)
```

```python
cv_scores_knn = cross_val_score(knn, X_scaled, y, cv=5, scoring='accuracy')
print(f"KNN Cross-Validation Scores: {cv_scores_knn}")
print(f"KNN CV Mean: {cv_scores_knn.mean():.4f}")
print(f"KNN CV Std:  {cv_scores_knn.std():.4f}")
```

## Model Comparison Table

```python
from sklearn.metrics import precision_score, recall_score, f1_score

results = {
    'Model': ['Logistic Regression', 'Decision Tree', 'KNN'],
    'Accuracy':  [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_knn)
    ],
    'Precision': [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_dt),
        precision_score(y_test, y_pred_knn)
    ],
    'Recall': [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_dt),
        recall_score(y_test, y_pred_knn)
    ],
    'F1-Score': [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_dt),
        f1_score(y_test, y_pred_knn)
    ],
    'CV Mean': [
        cv_scores_lr.mean(),
        cv_scores_dt.mean(),
        cv_scores_knn.mean()
    ],
    'CV Std': [
        cv_scores_lr.std(),
        cv_scores_dt.std(),
        cv_scores_knn.std()
    ]
}

df_results = pd.DataFrame(results)
df_results = df_results.round(4)
print(df_results.to_string(index=False))
```

## Written Interpretation

Based on the results in the comparison table, Logistic Regression provides the most reliable and balanced performance among the three models. While the differences in Accuracy between models are not extreme, Logistic Regression stands out due to its stronger F1-score and more stable Recall. In medical classification tasks, Recall is a critical metric because it reflects the model’s ability to correctly identify patients with diabetes. A higher Recall reduces the number of false negatives, which is especially important since undetected cases may lead to delayed treatment and serious complications.

One reason for this performance is the nature of the dataset. Key features such as Glucose and BMI show a relatively consistent and monotonic relationship with the target variable, which makes them well-suited for a linear decision boundary. Logistic Regression is able to capture this structure effectively without introducing unnecessary complexity.

In contrast, the Decision Tree model appears to suffer from overfitting. This is indicated not only by its slightly less stable test performance, but also by the higher variability observed in cross-validation scores. The model likely adapts too closely to patterns in the training data, which reduces its ability to generalize.

The KNN model performs reasonably well but shows less consistency overall. Despite applying feature scaling, its dependence on distance calculations and the fixed value of k makes it more sensitive to local data structure, which can lead to fluctuations in performance.

An important observation is that the Decision Tree achieves lower Recall for the positive class (Outcome = 1), meaning it fails to detect a larger portion of diabetic patients compared to Logistic Regression. From a practical perspective, this type of error is more critical than a false positive, which further supports choosing Logistic Regression as the preferred model in this case.
